## Basic Calculator

This calculator provides functions for basic arithmetic operations: addition, subtraction, multiplication, and division.

```python
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        return "Error! Division by zero."
    return x / y
```

```python
# Demonstration of the calculator functions

num1 = 10
num2 = 5

print(f"{num1} + {num2} = {add(num1, num2)}")
print(f"{num1} - {num2} = {subtract(num1, num2)}")
print(f"{num1} * {num2} = {multiply(num1, num2)}")
print(f"{num1} / {num2} = {divide(num1, num2)}")

# Test division by zero
num3 = 0
print(f"{num1} / {num3} = {divide(num1, num3)}")
```

## 글자수 계산기 (Character Counter)

이 계산기는 한국어 문장의 띄어쓰기를 포함한 총 글자수를 계산합니다.

```python
def count_characters(sentence):
    """
    주어진 문장의 총 글자수(띄어쓰기 포함)를 계산합니다.
    """
    return len(sentence)
```

```python
# 글자수 계산기 함수 시연

korean_sentence = "안녕하세요! 파이썬으로 글자수를 세어봅시다."
character_count = count_characters(korean_sentence)

print(f"문장: '{korean_sentence}'")
print(f"총 글자수 (띄어쓰기 포함): {character_count} 글자")

korean_sentence_2 = "구름 한 점 없는 파란 하늘이었다."
character_count_2 = count_characters(korean_sentence_2)

print(f"\n문장: '{korean_sentence_2}'")
print(f"총 글자수 (띄어쓰기 포함): {character_count_2} 글자")
```

```python
# 전체 결측값의 개수를 계산합니다.
total_missing = secom_df.isnull().sum().sum()

# 전체 데이터 포인트의 개수를 계산합니다.
total_elements = secom_df.size

# 결측값 비율을 계산합니다.
missing_percentage = (total_missing / total_elements) * 100

print(f"`secom_df`의 전체 결측값 개수: {total_missing}")
print(f"`secom_df`의 전체 요소 개수: {total_elements}")
print(f"`secom_df`의 전체 결측 비율: {missing_percentage:.2f}%")
```

```python
# 각 열별 결측값 개수를 계산합니다.
missing_per_column = secom_df.isnull().sum()

# 각 열별 결측 비율을 계산합니다.
missing_percentage_per_column = (missing_per_column / len(secom_df)) * 100

# 결측 비율이 높은 순서대로 정렬하여 출력합니다.
print("각 열별 결측 비율 (내림차순):\n")
display(missing_percentage_per_column.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 결측 비율이 높은 순서대로 정렬합니다.
missing_percentage_sorted = missing_percentage_per_column.sort_values(ascending=False)

# 시각화를 위한 데이터프레임을 생성합니다.
missing_df = pd.DataFrame({
    'Column Index': missing_percentage_sorted.index.astype(str),
    'Missing Percentage': missing_percentage_sorted.values
})

# 막대 그래프를 그립니다.
plt.figure(figsize=(15, 6))
sns.barplot(x='Column Index', y='Missing Percentage', data=missing_df.head(50), palette='viridis')
plt.title('상위 50개 열의 결측 비율')
plt.xlabel('열 인덱스')
plt.ylabel('결측 비율 (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# 각 열의 평균값으로 결측치를 채웁니다.
secom_df_mean = secom_df.fillna(secom_df.mean())

# 변경 사항을 확인하기 위해 새로운 데이터프레임의 상위 5행을 표시합니다.
display(secom_df_mean.head())
```

```python
# 결측값이 성공적으로 채워졌는지 확인하기 위해 secom_df_mean의 결측값 개수를 다시 확인합니다.
print("secom_df_mean의 전체 결측값 개수:", secom_df_mean.isnull().sum().sum())
```

```python
# 각 열의 중앙값으로 결측치를 채웁니다.
secom_df_med = secom_df.fillna(secom_df.median())

# 변경 사항을 확인하기 위해 새로운 데이터프레임의 상위 5행을 표시합니다.
display(secom_df_med.head())
```

```python
# 결측값이 성공적으로 채워졌는지 확인하기 위해 secom_df_med의 결측값 개수를 다시 확인합니다.
print("secom_df_med의 전체 결측값 개수:", secom_df_med.isnull().sum().sum())
```

```python
from sklearn.impute import KNNImputer
import numpy as np

# KNNImputer 인스턴스를 생성합니다. (n_neighbors는 k값)
imputer = KNNImputer(n_neighbors=5)

# secom_df에 KNN 대치를 적용합니다.
# KNNImputer는 numpy 배열을 반환하므로, 다시 DataFrame으로 변환해야 합니다.
secom_df_knn_array = imputer.fit_transform(secom_df)
secom_df_knn = pd.DataFrame(secom_df_knn_array, columns=secom_df.columns)

# 변경 사항을 확인하기 위해 새로운 데이터프레임의 상위 5행을 표시합니다.
display(secom_df_knn.head())
```

```python
# 결측값이 성공적으로 채워졌는지 확인하기 위해 secom_df_knn의 결측값 개수를 다시 확인합니다.
print("secom_df_knn의 전체 결측값 개수:", secom_df_knn.isnull().sum().sum())
```
