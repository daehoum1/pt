## 1. Introduction: GCN and Missing Node Features

Cora is a graph of documents with node features and class labels. In this lab, a portion of nodes has **no observed feature vector**. We study whether better feature imputation and a suitable number of GCN layers can improve node classification.

### 1.1 Graph Convolutional Network (GCN)

For adjacency matrix $A$, after adding self-loops and applying symmetric normalization, a GCN layer computes:

$$H^{(\ell+1)} = \sigma\!\left(\widetilde{D}^{-1/2}\widetilde{A}\widetilde{D}^{-1/2}H^{(\ell)}W^{(\ell)}+b^{(\ell)}\right).$$

Each node aggregates information from neighbors, then transforms that information for classification. The **provided Zero/Mean baselines use three GCN layers and 64 hidden units**. For your method, you may change **the number of GCN layers** with `num_layers`, while keeping hidden width 64 and the layer operation and activation functions unchanged.

### 1.2 Structural Missingness and Baseline Imputation

Let $O$ be the set of nodes with **observed feature vectors**. An entire feature vector is either observed or missing; an imputer must never read the hidden original values.

**Zero Imputation** fills every missing node-feature entry with zero:

$$\widehat X_{ij}=\begin{cases}X_{ij},&i\in O,\\0,&i\notin O.\end{cases}$$

**Mean Imputation** fills missing entries with the feature-wise mean of **observed nodes only**:

$$\mu_j=\frac{1}{|O|}\sum_{i\in O}X_{ij},\qquad
\widehat X_{ij}=\begin{cases}X_{ij},&i\in O,\\\mu_j,&i\notin O.\end{cases}$$

Implement a different imputation method using the **observed features and fixed graph**. Methods may use neighboring nodes, iterative propagation, or other justified strategies. You may also tune the **number of GCN layers**, learning rate, optimizer, dropout, weight decay, epochs, and early stopping. **Maximize Validation Accuracy**, then evaluate the frozen choices on Test. / **결측치 대체 방법, GCN 계층 수, 학습 하이퍼파라미터**를 조정하여 Validation Accuracy를 높이세요. ### 2.1 Load Cora and fix the graph

Load `Planetoid(Cora)`, keep only the largest connected component, make edges undirected, and add self-loops. The notebook constructs the **same normalized adjacency** for every method. Do not alter the graph or use hidden features in an imputation method.


```python
# PROVIDED CORA DATA LOADER — KEEP GRAPH FIXED
raw = Planetoid(root=str(DATA_ROOT), name=DATASET_NAME)[0]
raw_edge = raw.edge_index.cpu().numpy().astype(np.int64)
n_raw = raw.num_nodes
raw_graph = sp.coo_matrix(
    (np.ones(raw_edge.shape[1], dtype=np.int8), (raw_edge[0], raw_edge[1])),
    shape=(n_raw, n_raw),
).tocsr()
_, component_label = connected_components(raw_graph, directed=False)
lcc_label = np.argmax(np.bincount(component_label))
selected_nodes = np.flatnonzero(component_label == lcc_label)
node_map = np.full(n_raw, -1, dtype=np.int64)
node_map[selected_nodes] = np.arange(len(selected_nodes))
keep_edges = (node_map[raw_edge[0]] >= 0) & (node_map[raw_edge[1]] >= 0)
edges = node_map[raw_edge[:, keep_edges]]
# Undirected graph, without duplicate edges, with one self-loop per node.
edges = np.concatenate([edges, edges[::-1]], axis=1)
n_nodes = len(selected_nodes)
loops = np.stack([np.arange(n_nodes), np.arange(n_nodes)])
edges = np.unique(np.concatenate([edges, loops], axis=1), axis=1)

X_FULL = raw.x[selected_nodes].float().to(DEVICE)  # Use ONLY inside the fixed observed-data mask below.
Y_ALL = raw.y[selected_nodes].long().to(DEVICE)     # Test labels used ONLY in final evaluation.
EDGE_INDEX = torch.tensor(edges, dtype=torch.long, device=DEVICE)
N, D = X_FULL.shape
NUM_CLASSES = int(Y_ALL.max().item()) + 1

# Symmetric normalized adjacency for the fixed GCN (also available to custom imputers).
row, col = EDGE_INDEX[0], EDGE_INDEX[1]
degree = torch.bincount(row, minlength=N).to(DEVICE).float()
a_sym = (degree[row] * degree[col]).clamp_min(1.0).rsqrt()
A_SYM = torch.sparse_coo_tensor(EDGE_INDEX, a_sym, (N, N)).coalesce()
print(f"Dataset: {DATASET_NAME} / LCC, nodes={N}, features={D}, classes={NUM_CLASSES}")
print(f"Undirected edges including self-loops={EDGE_INDEX.size(1):,}")
assert N > 1500 and NUM_CLASSES == 7 and torch.isfinite(X_FULL).all()
```

### 2.2 Fixed splits and missingness masks — BEFORE imputation

Use the same graph and the same class-balanced Train/Validation/Test split for all methods under each fixed seed. Structural missingness is generated from a **fixed node-level random score**, so a node missing at a lower rate remains missing at a higher rate. The four displayed rates are nominal; the code prints the realized percentages.

**Do not modify these splits or masks.** Never give an imputer the true feature vector of a missing node. The class labels of Validation/Test nodes are excluded from all model weight updates.


```python
# FIXED SPLITS AND MISSINGNESS — DO NOT MODIFY
SPLITS = {}
OBSERVED_MASKS = {}
for seed in SEEDS:
    rng = np.random.RandomState(seed)
    dev = rng.choice(N, 1500, replace=False)
    # Examine only development labels; held-out Test labels are not needed.
    dev_labels = Y_ALL[torch.tensor(dev, device=DEVICE)].cpu().numpy()
    train_parts = []
    for cls in range(NUM_CLASSES):
        pool = dev[dev_labels == cls]
        if len(pool) < 20:
            raise RuntimeError(f"Seed {seed}: fewer than 20 development nodes in class {cls}")
        train_parts.append(rng.choice(pool, 20, replace=False))
    train_idx = np.concatenate(train_parts).astype(np.int64)
    val_idx = np.setdiff1d(dev, train_idx)
    test_idx = np.setdiff1d(np.arange(N), dev)
    assert len(train_idx) == 20 * NUM_CLASSES
    assert len(np.intersect1d(train_idx, val_idx)) == 0
    assert len(np.intersect1d(train_idx, test_idx)) == 0
    assert len(np.intersect1d(val_idx, test_idx)) == 0
    SPLITS[seed] = {
        "train": torch.tensor(train_idx, device=DEVICE, dtype=torch.long),
        "val": torch.tensor(val_idx, device=DEVICE, dtype=torch.long),
        "test": torch.tensor(test_idx, device=DEVICE, dtype=torch.long),
    }
    # Fixed node-level random scores; same ordering at every missing rate.
    mask_rng = np.random.default_rng(seed + 100000)
    node_scores = mask_rng.random(N)
    for rate in MISSING_RATES:
        observed_rows = node_scores >= rate
        if not observed_rows.any():
            observed_rows[np.argmax(node_scores)] = True
        mask = torch.tensor(observed_rows, dtype=torch.bool, device=DEVICE)
        OBSERVED_MASKS[(seed, rate)] = mask

rows = []
for seed in SEEDS:
    for rate in MISSING_RATES:
        mask = OBSERVED_MASKS[(seed, rate)]
        rows.append({"Seed": seed, "Missing rate (target)": rate,
                     "Realized missing": 1 - mask.float().mean().item(),
                     "Known nodes": int(mask.sum().item()),
                     "Train labels": len(SPLITS[seed]["train"]),
                     "Val labels": len(SPLITS[seed]["val"]),
                     "Test labels": len(SPLITS[seed]["test"])})
mask_summary = pd.DataFrame(rows)
display(mask_summary)
for seed in SEEDS:
    for r1, r2 in zip(MISSING_RATES[:-1], MISSING_RATES[1:]):
        assert torch.all(OBSERVED_MASKS[(seed, r2)] <= OBSERVED_MASKS[(seed, r1)])
```

Do not change the layer type, hidden width, aggregation or activation.

The code trains with **Train labels only** and selects the best epoch using **Validation Accuracy**. Training hyperparameters and the layer count may change for your method; baseline settings stay fixed.


```python
# PROVIDED GCN LAYER DESIGN AND TRAINING HELPERS
# Baselines use 3 layers; custom experiments can tune num_layers.
# GCN aggregation, hidden width, and activation stay fixed.
GCN_NUM_LAYERS = 3  # Default and baseline depth
GCN_HIDDEN_DIM = 64

class GCNClassifier(nn.Module):
    def __init__(self, input_dim, n_classes, dropout=0.5, num_layers=GCN_NUM_LAYERS):
        super().__init__()
        if isinstance(num_layers, bool) or not isinstance(num_layers, (int, np.integer)) or num_layers < 1:
            raise ValueError("num_layers must be a positive integer")
        self.num_layers = int(num_layers)
        dims = [input_dim] + [GCN_HIDDEN_DIM] * (self.num_layers - 1) + [n_classes]
        self.layers = nn.ModuleList([
            nn.Linear(dims[i], dims[i + 1]) for i in range(self.num_layers)
        ])
        self.dropout = float(dropout)
        if not 0.0 <= self.dropout < 1.0:
            raise ValueError("dropout must be in [0, 1)")
        for layer in self.layers:
            nn.init.xavier_uniform_(layer.weight)
            nn.init.zeros_(layer.bias)

    def forward(self, x, adj):
        for index, layer in enumerate(self.layers):
            x = layer(torch.sparse.mm(adj, x))
            if index != len(self.layers) - 1:
                x = F.relu(x)
                x = F.dropout(x, p=self.dropout, training=self.training)
        return x  # node-class logits [N, NUM_CLASSES]

def make_gcn(config):
    if "hidden_dim" in config or "layers" in config:
        raise ValueError("GCN layer type and hidden dimension are fixed; only num_layers is adjustable")
    return GCNClassifier(
        D, NUM_CLASSES,
        dropout=config["dropout"],
        num_layers=config.get("num_layers", GCN_NUM_LAYERS),
    ).to(DEVICE)

@torch.no_grad()
def subset_accuracy(logits, labels, idx):
    return (logits[idx].argmax(dim=1) == labels[idx]).float().mean().item()

def fit_one_gcn(x_input, seed, config):
    # Paired initialization across methods for the same seed.
    set_seed(seed + 20000)
    model = make_gcn(config)
    opt_name = str(config.get("optimizer", "adam")).lower()
    lr = float(config["lr"])
    weight_decay = float(config["weight_decay"])
    if opt_name == "adam":
        optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    elif opt_name == "adamw":
        optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    elif opt_name == "sgd":
        optimizer = torch.optim.SGD(model.parameters(), lr=lr, weight_decay=weight_decay,
                                    momentum=float(config.get("momentum", 0.9)))
    else:
        raise ValueError("optimizer must be 'adam', 'adamw' or 'sgd'")

    split = SPLITS[seed]
    train_idx, val_idx = split["train"], split["val"]
    best_val, best_loss, best_epoch, best_state = -1.0, float("inf"), -1, None
    stale = 0
    history = []
    for epoch in range(1, int(config["epochs"]) + 1):
        model.train()
        optimizer.zero_grad()
        logits = model(x_input, A_SYM)
        loss = F.cross_entropy(logits[train_idx], Y_ALL[train_idx])
        if not torch.isfinite(loss):
            raise FloatingPointError("Training loss is not finite")
        loss.backward()
        if config.get("grad_clip") is not None:
            nn.utils.clip_grad_norm_(model.parameters(), float(config["grad_clip"]))
        optimizer.step()

        model.eval()
        with torch.no_grad():
            logits_val = model(x_input, A_SYM)
            val_loss = F.cross_entropy(logits_val[val_idx], Y_ALL[val_idx]).item()
            val_acc = subset_accuracy(logits_val, Y_ALL, val_idx)
        history.append({"epoch": epoch, "train_loss": float(loss.item()),
                        "val_loss": val_loss, "val_acc": val_acc})
        # Tie-break with Validation loss; never use Test scores here.
        improved = (val_acc > best_val + 1e-9 or
                    (abs(val_acc - best_val) <= 1e-9 and val_loss < best_loss - 1e-9))
        if improved:
            best_val, best_loss, best_epoch = val_acc, val_loss, epoch
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            stale = 0
        else:
            stale += 1
        if stale >= int(config["patience"]):
            break
    assert best_state is not None
    model.load_state_dict(best_state)
    model.eval()
    return model, best_epoch, best_val, history

# Fast smoke tests for baseline (3 layers) and a custom depth (4 layers).
# No Test labels are used.
_smoke_cfg = dict(dropout=0.2, lr=0.005, weight_decay=0.0005,
                  epochs=2, patience=2, optimizer="adam")
_smoke_x = get_input_features(SEEDS[0], MISSING_RATES[0], {"method": "zero"})
_smoke_model, _smoke_ep, _smoke_val, _ = fit_one_gcn(_smoke_x, SEEDS[0], _smoke_cfg)
assert len(_smoke_model.layers) == GCN_NUM_LAYERS and _smoke_model.layers[0].out_features == 64
assert 0.0 <= _smoke_val <= 1.0
_smoke_4_cfg = {**_smoke_cfg, "num_layers": 4}
_smoke_4_model, _, _smoke_4_val, _ = fit_one_gcn(_smoke_x, SEEDS[0], _smoke_4_cfg)
assert len(_smoke_4_model.layers) == 4 and _smoke_4_model.layers[-1].out_features == NUM_CLASSES
assert 0.0 <= _smoke_4_val <= 1.0
print("GCN depth (3 / 4 layers), forward/backward, and validation check: PASSED")
```

### 3.3 Provided baselines: Zero and Mean Imputation

Both baselines have **identical fixed training settings**. Their only difference is how missing features are filled. Evaluate both on all **4 missing rates × 3 seeds**, saving the best Validation checkpoint for each condition.

**Do not tune baseline settings.** The provided baseline code does **not** access Test labels. / 두 Baseline의 학습 설정은 고정이며, 성능 비교는 Validation으로만 수행합니다.


```python
# FIXED BASELINE SETTINGS — DO NOT MODIFY (implicit num_layers=3)
BASELINE_GCN = dict(dropout=0.5, lr=0.005, weight_decay=0.0,
                    optimizer="adam", epochs=250, patience=50)
BASELINES = {
    "Zero Imputation": {"method": "zero", **BASELINE_GCN},
    "Mean Imputation": {"method": "mean", **BASELINE_GCN},
}
CONDITIONS = [(seed, rate) for rate in MISSING_RATES for seed in SEEDS]
assert len(CONDITIONS) == len(SEEDS) * len(MISSING_RATES) == 12
assert len(BASELINES) == 2
```

```python
# TRAIN BOTH BASELINES — VALIDATION ONLY
# This cell NEVER evaluates Test labels.
checkpoint_files = []
validation_records = []
learning_records = []

def checkpoint_name(method_label, rate, seed):
    safe = method_label.lower().replace(" ", "_")
    rate_label = str(rate).replace(".", "p")
    return OUTPUT_ROOT / f"{safe}_missing_{rate_label}_seed_{seed}.pt"

def mask_signature(seed, rate):
    mask = OBSERVED_MASKS[(seed, rate)].cpu().numpy()
    return hashlib.sha256(mask.tobytes()).hexdigest()

def feature_signature(x):
    """Verify that the final evaluation recomputes the SAME imputed features."""
    return hashlib.sha256(x.detach().cpu().contiguous().numpy().tobytes()).hexdigest()

def train_and_save(label, cfg, seed, rate):
    x_input = get_input_features(seed, rate, cfg)
    model, best_epoch, best_acc, history = fit_one_gcn(x_input, seed, cfg)
    path = checkpoint_name(label, rate, seed)
    torch.save({
        "state_dict": {k: v.cpu() for k, v in model.state_dict().items()},
        "config": copy.deepcopy(cfg), "seed": seed, "missing_rate": rate,
        "label": label, "best_epoch": best_epoch,
        "best_validation_accuracy": best_acc,
        "mask_sha256": mask_signature(seed, rate),
        "features_sha256": feature_signature(x_input),
    }, path)
    checkpoint_files.append(path)
    validation_records.append({"Method": label, "Missing rate": rate, "Seed": seed,
                               "Best epoch": best_epoch, "Best Validation Accuracy": best_acc,
                               "GCN Layers": int(cfg.get("num_layers", GCN_NUM_LAYERS)),
                               "Checkpoint": str(path)})
    learning_records.extend([{**entry, "Method": label, "Missing rate": rate,
                              "Seed": seed} for entry in history])
    print(f"{label:>20} | missing={rate:>5.1%} seed={seed}: "
          f"val={best_acc*100:5.2f}% at epoch {best_epoch}")

for label, config in BASELINES.items():
    for seed, rate in CONDITIONS:
        train_and_save(label, config, seed, rate)

baseline_val_table = pd.DataFrame(validation_records)
display(baseline_val_table.groupby(["Method", "Missing rate"])
        ["Best Validation Accuracy"].agg(["mean", "std"]).round(4))
```

## 5. Final Test Evaluation — DO NOT MODIFY

Finish your imputation experiments and freeze all configurations and **Validation-best checkpoints** before running this cell. There must be **36 checkpoints**: Zero, Mean, and Your Imputation × four missing rates × three seeds. The final Test cell checks saved configurations and feature signatures and must be run **once only**. Do not tune based on any Test result.

For missing rate $r$, let $A_{r,m}$ be the **mean Test Accuracy (%)** over three seeds for method $m$. The stronger baseline at that rate is:

$$B_r=\max\!\left(A_{r,\mathrm{Zero}},\;A_{r,\mathrm{Mean}}\right).$$

The average accuracy improvement in **percentage points** is:

$$\Delta_r=A_{r,\mathrm{Yours}}-B_r,\qquad
\boxed{\mathrm{Average\ Improvement\ (pp)}=\frac{1}{4}\sum_{r\in\{0.10,0.50,0.90,0.995\}}\Delta_r}.$$

A negative value means your method is weaker than the stronger baseline on average. Also report your own average Test Accuracy over all four rates. **These final Test quantities must never be used to choose the imputation method or its hyperparameters.**


```python
# FINAL TEST EVALUATION — DO NOT MODIFY / DO NOT USE FOR MODEL SELECTION.
# Run ONCE after fixing all settings and saving all 36 checkpoints.
assert len(checkpoint_files) == len(CONDITIONS) * 3
final_records = []
seen = set()
for path in checkpoint_files:
    saved = torch.load(path, map_location="cpu", weights_only=True)
    label = saved["label"]
    seed = int(saved["seed"])
    rate = float(saved["missing_rate"])
    key = (label, seed, rate)
    assert key not in seen, f"Duplicate checkpoint for {key}"
    seen.add(key)
    assert saved["mask_sha256"] == mask_signature(seed, rate), "Missing mask changed"
    if label in BASELINES:
        assert saved["config"] == BASELINES[label], "Fixed baseline configuration was changed"
    else:
        assert label == "Your Imputation"
        assert saved["config"] == OPTIMIZED_CONFIGS[rate], "Custom config differs from saved checkpoint"
    config = saved["config"]
    assert all(k not in config for k in ("hidden_dim", "layers")), "GCN layer design was changed"
    assert isinstance(config.get("num_layers", GCN_NUM_LAYERS), int) and config.get("num_layers", GCN_NUM_LAYERS) >= 1
    if label in BASELINES:
        assert config.get("num_layers", GCN_NUM_LAYERS) == GCN_NUM_LAYERS, "Baseline depth was changed"
    x_input = get_input_features(seed, rate, config)
    assert feature_signature(x_input) == saved["features_sha256"], (
        "Imputed features changed since training. Restore the chosen imputer "
        "and re-train BEFORE examining Test results."
    )
    model = make_gcn(config)
    model.load_state_dict(saved["state_dict"])
    model.eval()
    with torch.no_grad():
        logits = model(x_input, A_SYM)
        # Test labels are accessed ONLY here.
        test_acc = subset_accuracy(logits, Y_ALL, SPLITS[seed]["test"])
    final_records.append({"Missing rate": rate, "Seed": seed, "Method": label,
                          "Test Accuracy (%)": test_acc * 100,
                          "Best Validation Accuracy (%)": saved["best_validation_accuracy"] * 100,
                          "Best epoch": saved["best_epoch"],
                          "GCN Layers": int(config.get("num_layers", GCN_NUM_LAYERS))})
assert len(seen) == len(CONDITIONS) * 3
assert all((method, seed, rate) in seen for method in
           list(BASELINES) + ["Your Imputation"] for seed, rate in CONDITIONS)
final_df = pd.DataFrame(final_records)
final_summary = (final_df.groupby(["Missing rate", "Method"])["Test Accuracy (%)"]
                 .agg(["mean", "std"]).reset_index())
print("FINAL TEST ACCURACY (%) — mean ± sample std across three seeds")
final_display = final_summary.copy()
final_display["Missing rate"] = final_display["Missing rate"].map(lambda x: f"{x:.1%}")
display(final_display.round(2))

wide = final_summary.pivot(index="Missing rate", columns="Method", values="mean")
wide["Best baseline"] = wide[["Zero Imputation", "Mean Imputation"]].max(axis=1)
wide["Improvement (pp)"] = wide["Your Imputation"] - wide["Best baseline"]
print("Improvement over the stronger baseline (percentage points)")
wide_display = wide.copy()
wide_display.index = [f"{r:.1%}" for r in wide_display.index]
display(wide_display.round(2))
score = wide["Improvement (pp)"].mean()
mean_acc = wide["Your Imputation"].mean()
print(f"FINAL AVERAGE TEST ACCURACY (YOUR METHOD): {mean_acc:.3f}%")
print(f"FINAL AVERAGE TEST ACCURACY IMPROVEMENT: {score:+.3f} percentage points")

fig, ax = plt.subplots(figsize=(8, 4.5))
for method, subset in final_summary.groupby("Method"):
    subset = subset.sort_values("Missing rate")
    ax.errorbar(subset["Missing rate"] * 100, subset["mean"],
                yerr=subset["std"].fillna(0), marker="o", capsize=3, label=method)
ax.set_xlabel("Missing features (%)")
ax.set_ylabel("Test Accuracy (%)")
ax.set_title("GCN node classification with missing features")
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
# These results are final. Never use Test performance to choose settings.
```
