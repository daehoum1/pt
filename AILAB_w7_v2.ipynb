## 1. Introduction: DCGAN for Face Generation

### 1.1 Generative Adversarial Networks (GANs)

A **Generator** $G$ transforms random latent noise $z$ into a synthetic image $G(z)$. A **Discriminator** $D$ estimates the probability that an image is real.

```text
Random noise z (N, 100, 1, 1)
               |
          Generator G
               |
        Fake face (64x64) ----> Discriminator D ----> Real / Fake
                                     ^
                                     |
                              Real CelebA face
```

The two networks compete during training. The baseline uses **Binary Cross-Entropy (BCE)** and alternates Discriminator and Generator updates:

- **Discriminator:** classify real images as real and generated images as fake.
- **Generator:** make generated images more likely to be classified as real.

A basic non-saturating GAN objective is

$$
L_D = -\mathbb{E}_{x\sim p_{\mathrm{data}}}\log D(x)
              -\mathbb{E}_{z}\log(1-D(G(z))),
$$

$$
L_G = -\mathbb{E}_{z}\log D(G(z)).
$$

### 1.2 Why evaluate FID?

Generator and Discriminator losses measure **training objectives**; neither directly measures the quality of generated images. The **Fréchet Inception Distance (FID)** compares the distributions of features extracted from real and generated images. **Lower FID indicates closer feature distributions**, under a common evaluation protocol.

**Discriminative model D**

- Xavier Initialization.
- Real data or fake data for input (size = 64 × 64).
- Its structure is the reverse version of the generative model G where the upconvolutions are replaced with convolutions and the last output size is changed to **(bs, 1, 1, 1)**.
- Use the batch normalization except for the last layer.

> **Implementation note:** The original guide uses the word *stride* for the change from 5 to 4. In the fixed layer specifications below, however, **4 is the convolution kernel size**. The Generator uses stride 1 in its first block and stride 2 in its remaining blocks. Use the fixed layer specifications below when implementing the model.

### 3. Loss function

**Discriminative model D**

- Binary cross entropy loss for the real data.
- Binary cross entropy loss for the fake data.

For real data:

$$
L_{D,\mathrm{real}}
= -\frac{1}{N}\sum_{i=1}^{N}\left[
 y_i^{\mathrm{real}}\log D(x_i)
 +(1-y_i^{\mathrm{real}})\log\bigl(1-D(x_i)\bigr)
\right]
$$

For fake data:

$$
L_{D,\mathrm{fake}}
= -\frac{1}{N}\sum_{i=1}^{N}\left[
 y_i^{\mathrm{fake}}\log D(G(z_i))
 +(1-y_i^{\mathrm{fake}})\log\bigl(1-D(G(z_i))\bigr)
\right]
$$

where $y_i^{\mathrm{real}} = 1$ and $y_i^{\mathrm{fake}} = 0$.

**Generative model G**

- Binary cross entropy loss for the real data (it is actually fake).

$$
L_{G}
= -\frac{1}{N}\sum_{i=1}^{N}\left[
 y_i^{\mathrm{real}}\log D(G(z_i))
 +(1-y_i^{\mathrm{real}})\log\bigl(1-D(G(z_i))\bigr)
\right]
$$

where $y_i^{\mathrm{real}} = 1$.

> **Notation:** $x_i$ denotes a real image, while $z_i$ denotes a latent vector. Thus, the discriminator output for a real image is written as $D(x_i)$.

### 4. Training Strategy

- Same as the GAN training algorithm.
- The generative model G and the discriminative model D are trained alternatively during training.
- When the parameters of model D are being updated, back-propagation in model G is blocked.

## 3. Implement the DCGAN Models [80 points]

### 3.1 Provided Convolution Helpers

Use the following `conv2d()` and `upconv2d()` functions. The provided helpers use **Xavier uniform initialization** as a default. You may experiment with initialization as a **training setting**, but the layer structure in Section 2 must not change.

````python
# STEP 5: PROVIDED CONVOLUTION HELPERS
# DO NOT CHANGE THE LAYER TYPES OR ARCHITECTURE.
def conv2d(channel_in, channel_out, kernel_size, stride, padding,
           activation, batch_norm=True):
    layers = [nn.Conv2d(channel_in, channel_out, kernel_size, stride,
                        padding, bias=False)]
    nn.init.xavier_uniform_(layers[0].weight)
    if batch_norm:
        layers.append(nn.BatchNorm2d(channel_out))
    if activation == "ReLU":
        layers.append(nn.ReLU(inplace=True))
    elif activation == "LeakyReLU":
        layers.append(nn.LeakyReLU(0.2, inplace=True))
    elif activation == "Tanh":
        layers.append(nn.Tanh())
    elif activation == "Sigmoid":
        layers.append(nn.Sigmoid())
    return nn.Sequential(*layers)


def upconv2d(channel_in, channel_out, kernel_size, stride, padding,
             activation, batch_norm=True):
    layers = [nn.ConvTranspose2d(channel_in, channel_out, kernel_size,
                                 stride, padding, bias=False)]
    nn.init.xavier_uniform_(layers[0].weight)
    if batch_norm:
        layers.append(nn.BatchNorm2d(channel_out))
    if activation == "ReLU":
        layers.append(nn.ReLU(inplace=True))
    elif activation == "LeakyReLU":
        layers.append(nn.LeakyReLU(0.2, inplace=True))
    elif activation == "Tanh":
        layers.append(nn.Tanh())
    elif activation == "Sigmoid":
        layers.append(nn.Sigmoid())
    return nn.Sequential(*layers)
````

### 4.2 Alternating Generator / Discriminator Training

The code below is a working baseline for DCGAN training.

1. Update the Discriminator with **real** and **detached fake** images.
2. Update the Generator by asking the Discriminator to classify fake images as real.
3. Record the average loss per epoch.

Feel free to experiment with your training schedule, optimizer, or augmentation. Keep the architecture and the final FID protocol unchanged. To reduce notebook size, this loop **does not save an image grid for every epoch**.

````python
# STEP 8: TRAIN DCGAN
# Baseline implementation. Training settings and update schedule are editable.
for epoch in range(NUM_EPOCHS):
    model_G.train()
    model_D.train()
    epoch_loss_g = 0.0
    epoch_loss_d = 0.0
    num_batches = 0

    for real_images, _ in train_loader:
        real_images = real_images.to(device)
        batch_size = real_images.size(0)
        real_targets = torch.ones(batch_size, device=device)
        fake_targets = torch.zeros(batch_size, device=device)
        noise = torch.randn(batch_size, Z_SIZE, 1, 1, device=device)

        # (1) Discriminator update
        optimizer_D.zero_grad(set_to_none=True)
        generated_images = model_G(noise)
        real_scores = model_D(real_images).view(-1)
        fake_scores = model_D(generated_images.detach()).view(-1)
        loss_d = criterion(real_scores, real_targets) + criterion(fake_scores, fake_targets)
        loss_d.backward()
        optimizer_D.step()

        # (2) Generator update
        optimizer_G.zero_grad(set_to_none=True)
        generator_scores = model_D(generated_images).view(-1)
        loss_g = criterion(generator_scores, real_targets)
        loss_g.backward()
        optimizer_G.step()

        epoch_loss_d += float(loss_d.item())
        epoch_loss_g += float(loss_g.item())
        num_batches += 1

    avg_d = epoch_loss_d / num_batches
    avg_g = epoch_loss_g / num_batches
    train_loss_D.append(avg_d)
    train_loss_G.append(avg_g)
    print(f"Epoch {epoch + 1:3d}/{NUM_EPOCHS} | Loss_D: {avg_d:.4f} | Loss_G: {avg_g:.4f}")

assert len(train_loss_G) == NUM_EPOCHS and len(train_loss_D) == NUM_EPOCHS
````

````python
# ============================================================
# FINAL FID EVALUATION — DO NOT MODIFY THIS CELL
# ============================================================
assert "model_G" in globals() and "model_D" in globals()
assert len(train_loss_G) > 0 and len(train_loss_D) > 0, "Train the DCGAN first."
assert "_final_fid_evaluated" not in globals(), (
    "Final FID has already been evaluated in this session."
)

from torchmetrics.image.fid import FrechetInceptionDistance

FID_NUM_IMAGES = 5000
FID_BATCH_SIZE = 32
FID_SEED = 42

# Use the untouched evaluation transform, not a training augmentation.
reference_dataset = dset.ImageFolder(
    root=str(DATA_ROOT), transform=FID_REFERENCE_TRANSFORM
)
assert len(reference_dataset) >= FID_NUM_IMAGES, (
    "The fixed evaluation requires at least 5,000 real CelebA images."
)

reference_indices = torch.randperm(
    len(reference_dataset), generator=torch.Generator().manual_seed(FID_SEED)
)[:FID_NUM_IMAGES].tolist()
reference_loader = torch.utils.data.DataLoader(
    torch.utils.data.Subset(reference_dataset, reference_indices),
    batch_size=FID_BATCH_SIZE,
    shuffle=False,
    drop_last=False,
    num_workers=2,
    pin_memory=(device.type == "cuda"),
)

# TorchMetrics FID(normalize=False) expects RGB uint8 in [0, 255].
def _to_fid_uint8(images):
    return ((images.clamp(-1, 1) + 1.0) * 127.5).round().to(torch.uint8)

fid_metric = FrechetInceptionDistance(feature=2048, normalize=False)
fid_metric = fid_metric.set_dtype(torch.float64).to(device)
latent_rng = torch.Generator().manual_seed(FID_SEED)

was_training = model_G.training
model_G.eval()
try:
    with torch.no_grad():
        for real_images, _ in reference_loader:
            real_images = real_images.to(device)
            n = real_images.size(0)
            noise = torch.randn(n, Z_SIZE, 1, 1, generator=latent_rng).to(device)
            fake_images = model_G(noise)
            fid_metric.update(_to_fid_uint8(real_images), real=True)
            fid_metric.update(_to_fid_uint8(fake_images), real=False)
    final_fid = float(fid_metric.compute().item())
finally:
    model_G.train(was_training)

checkpoint_path = "DCGAN_final.pth"
torch.save({
    "generator_state_dict": model_G.state_dict(),
    "discriminator_state_dict": model_D.state_dict(),
    "z_size": Z_SIZE,
    "gfc": GFC,
    "dfc": DFC,
    "img_size": IMG_SIZE,
    "training_epochs": len(train_loss_G),
    "final_fid": final_fid,
    "fid_num_images": FID_NUM_IMAGES,
    "fid_seed": FID_SEED,
}, checkpoint_path)
_final_fid_evaluated = True
print("=" * 72)
print("DCGAN — FINAL FID RESULT")
print("=" * 72)
print(f"Final FID                   : {final_fid:.4f} (lower is better)")
print(f"Real / generated images     : {FID_NUM_IMAGES:,} / {FID_NUM_IMAGES:,}")
print(f"FID random seed             : {FID_SEED}")
print(f"Feature extractor           : Inception-v3 (2048 dimensions)")
print(f"Saved checkpoint            : {checkpoint_path}")
print("=" * 72)
````
